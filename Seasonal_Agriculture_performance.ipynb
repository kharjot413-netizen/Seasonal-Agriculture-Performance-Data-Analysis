# 5. Data Quality Analysis

```python
missing = df.isnull().sum().sort_values(ascending=False)
missing_percent = (df.isnull().mean()*100).sort_values(ascending=False)

missing_summary = pd.DataFrame({
    "Missing_Count": missing,
    "Missing_Percentage": missing_percent.round(2)
})
missing_summary[missing_summary["Missing_Count"] > 0]
```

```python
# Visualize missing values

plt.figure(figsize=(12, 5))
missing_plot = missing[missing > 0]
if len(missing_plot) > 0:
    sns.barplot(x=missing_plot.index, y=missing_plot.values)
    plt.xticks(rotation=60, ha="right")
    plt.xlabel("Column")
    plt.ylabel("Number of Missing Values")
    plt.title("Missing Values by Column")
    plt.tight_layout()
    plt.show()
else:
    print("No missing values found")
```

# 9. Outlier Analysis

# Univariate: Season distribution(Pie Chart)

```python
plt.figure(figsize=(7, 7))
df["Season"].value_counts().plot.pie(autopct='%1.1f%%', startangle=90, cmap='viridis')
plt.title("Distribution of Records by Season")
plt.ylabel("")
plt.tight_layout()
plt.show()
```

```python
# IQR-based outlier summary for numerical columns

outlier_summary = []

for col in numeric_columns:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    count = ((df[col] < lower) | (df[col] > upper)).sum()

    outlier_summary.append({
        "Column": col,
        "Lower_Bound": lower,
        "Upper_Bound": upper,
        "Outlier_Count": count,
        "Outlier_Percentage": round(count / len(df) * 100, 2)
    })

outlier_summary_df = pd.DataFrame(outlier_summary).sort_values(
    "Outlier_Count", ascending=False
    )
outlier_summary_df

```

# 12.  Seasonal Comparison

```python
# General seasonal summary

season_metrics = df.groupby("Season").agg(
    Records = ("Farm_ID", "count"),
    Average_Yield = ("Yield_Tonnes_Ha", "mean"),
    Total_Production = ("Production_Tonnes", "sum"),
    Average_Profit = ("Profit_INR", "mean"),
    Total_Profit=("Profit_INR", "sum"),
    Average_Water_Used = ("Water_Used_m3", "mean")
).round(2)

season_metrics
```

```python
# Compare crop performance within each season

crop_season_summary = (
    df.groupby(["Season", "Crop"])
    .agg(
        Average_Yield=("Yield_Tonnes_Ha", "mean"),
        Average_Profit=("Profit_INR", "mean"),
        Average_Water_Efficiency=("Water_Efficiency_t_per_1000m3", "mean"),
    )
    .round(2)
    .sort_values(["Season", "Average_Yield"], ascending=[True, False])
)
crop_season_summary
```

```python
# Seasonal Yield vs Rainfall Trend

plt.figure(figsize=(8,5))
sns.lineplot(data=df, x="Rainfall_mm", y="Yield_Tonnes_Ha", hue="Season", marker="o")
plt.title("Seasonal Yield vs Rainfall Trend")
plt.xlabel("Rainfall (mm)")
plt.ylabel("Yield (Tonnes/Ha)")
plt.legend(title="Season")
plt.tight_layout()
plt.show()
```

```python
# Seasonal Profitability and cost efficiency

season_profit_cost = df.groupby("Season").agg(
    Avg_Cost=("Total_Cost_INR", "mean"),
    Avg_Revenue=("Revenue_INR", "mean"),
    Avg_Profit=("Profit_INR", "mean")
).round(2)

season_profit_cost.plot(kind="bar", figsize=(8,5))
plt.title("Profitability vs Cost Efficiency by Season")
plt.ylabel("INR")
plt.tight_layout()
plt.show()
```

```python
# Seasonal Input usage Efficiency

# Group by season and calculate averages
season_inputs_eff = df.groupby("Season").agg(
    Fertilizer=("Fertilizer_kg_ha", "mean"),
    Pesticide=("Pesticide_Litre_ha", "mean"),
    Water=("Water_Used_m3", "mean"),
    Yield=("Yield_Tonnes_Ha", "mean")
).round(2)

# Calculate yield per total input
season_inputs_eff["Yield_per_Input"] = season_inputs_eff["Yield"] / (
    season_inputs_eff["Fertilizer"] + season_inputs_eff["Pesticide"] + season_inputs_eff["Water"]
)

print(season_inputs_eff)
```
